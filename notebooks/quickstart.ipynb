# Tutorial: remote-circuit Quickstart

**Audience:** lab members starting a circuit study.

**Prerequisites:** Python 3.12, `uv sync`, and a reviewed circuit YAML.

**Learning goal:** validate, run, inspect, and save one reproducible experiment without modifying the shared library.


## Outline

1. Load and validate the experiment
2. Run the smoke calculation
3. Inspect provenance and validity
4. Save an immutable result


```python
from pathlib import Path

from remote_circuit import load_experiment, run_experiment, validate_experiment

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
spec = load_experiment(ROOT / 'experiment.toml')
validate_experiment(spec)
spec.name, spec.model.type, spec.methods.enabled
```

## Run the smoke experiment

Smoke settings test the workflow; they are not evidence of numerical convergence.


```python
result = run_experiment(spec)
{
    'coordinates': {key: value.shape for key, value in result.coordinates.items()},
    'validity': {key: bool(value.all()) for key, value in result.diagnostics['validity'].items()},
    'git_commit': result.metadata['git_commit'],
    'git_dirty': result.metadata['git_dirty'],
}
```

## Save and extend

Result files never overwrite an existing run. For the exercise, change the sweep maximum in `experiment.toml`, choose a new filename, and compare `zz_z_khz`. A common mistake is interpreting smoke output as converged; move to `standard` and then `converged` before publication.


```python
destination = ROOT / 'results' / 'quickstart.h5'
if destination.exists():
    print(f'Already exists (immutable): {destination}')
else:
    print(result.save(destination))
```
